## 1. Executive Summary

**Objective:** Identify which customers are likely to churn, so the bank can take proactive retention action.

**Overall churn rate:** 16.1% of customers (1,627 of 10,127) have churned.

**Key finding:** Churn risk is driven almost entirely by customer *behaviour*, not demographics. Age, marital status, and gender show little to no meaningful relationship with churn. Income shows a weak, non-linear (U-shaped) relationship and isn't a reliable standalone predictor.

**The clearest risk indicators, in order of strength:**
1. **Declining transaction activity** — churned customers transact ~40% less than active customers (median 43 vs 70 transactions/year). This is the single strongest signal.
2. **Mid-range activity zone** — churn concentrates specifically among customers with 30-90 transactions and £1,000-£10,000 spend. High-activity customers (100+ transactions) rarely churn at all.
3. **Lower revolving balance & utilization ratio** — churned customers consistently show lower balances and utilization under 0.3.
4. **Rising contact count** — churned customers contact the bank more often (median 3 vs 2), while time spent inactive shows little difference between groups.


**Scope note:** This analysis is descriptive and correlation-based, using pandas, matplotlib, seaborn, and plotly only. No machine learning or predictive modelling was used, per project scope — findings represent observed patterns and risk indicators rather than statistically validated predictions.

## 2. Recommendations to Reduce Churn

* **Early-Warning System**: Flag accounts when transaction count drops below 50/year or falls >20% quarter-over-quarter.
* **Targeted Re-Engagement**: Offer cashback/rewards incentives to accounts showing declining card usage before they churn.
* **Proactive Support Outreach**: Track high-contact accounts (+0.20 correlation with churn) and offer retention incentives during those calls.
* **Promote Utilization**: Encourage low-balance users to set up recurring payments to keep accounts active.
* **Monitor the "Danger Zone"**: Flag accounts with 30-90 transactions and £1,000-£10,000 spend — this segment shows the highest churn concentration, while 100+ transaction customers are low-risk.
