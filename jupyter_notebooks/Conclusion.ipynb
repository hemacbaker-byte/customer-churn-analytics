## 1. Executive Summary

**_Objective_:** Identify which customers are likely to churn, so the bank can take proactive retention action.

**_Overall churn rate_:** 16.1% of customers (1,627 of 10,127) have churned.

**_Key finding_:** Churn risk is driven almost entirely by customer *behaviour*, not demographics. Age, marital status, and gender show little to no meaningful relationship with churn. Income shows a weak, non-linear (U-shaped) relationship and isn't a reliable standalone predictor.

**The clearest risk indicators, in order of strength:**
1. **_Declining transaction activity_** — churned customers transact ~40% less than active customers (median 43 vs 70 transactions/year). This is the single strongest signal.

2. **_Mid-range activity zone_** — churn concentrates specifically among customers with 30-90 transactions and £1,000-£10,000 spend. High-activity customers (100+ transactions) rarely churn at all.

3. **_Lower revolving balance & utilization ratio_** — churned customers consistently show lower balances and utilization under 0.3.

4. **_Rising contact count_** — churned customers contact the bank more often (median 3 vs 2), while time spent inactive shows little difference between groups.


**_Scope note_:** This analysis is descriptive and correlation-based, using pandas, matplotlib, seaborn, and plotly only. No machine learning or predictive modelling was used, per project scope — findings represent observed patterns and risk indicators rather than statistically validated predictions.

## 2. Recommendations to Reduce Churn

* **_Early-Warning System_**: Flag accounts when transaction count drops below 50/year or falls >20% quarter-over-quarter.

* **_Targeted Re-Engagement_**: Offer cashback/rewards incentives to accounts showing declining card usage before they churn.

* **_Proactive Support Outreach_**: Track high-contact accounts (+0.20 correlation with churn) and offer retention incentives during those calls.

* **_Promote Utilization_**: Encourage low-balance users to set up recurring payments to keep accounts active.

* **_Monitor the "Danger Zone"_**: Flag accounts with 30-90 transactions and £1,000-£10,000 spend — this segment shows the highest churn concentration, while 100+ transaction customers are low-risk.
